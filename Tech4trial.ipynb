```python
# x1=number of batches of product 1 per week, x2: number of batches of product 2 per week 
# WYNDOR EXAMPLE (SLIDE 6/15)
import pyomo.environ as pyo
solver_name= 'appsi_highs'
SOLVER= pyo.SolverFactory(solver_name)
assert SOLVER.available(), f'Solver {solver_name} is not available'

model=pyo.ConcreteModel('Wyndor Glass Co.')
#the string passed to the constructer will be the name of the model
#now, adding variables with the bounds
model.x1 = pyo.Var(bounds=(0,4))
model.x2 = pyo.Var(bounds=(0,6))

# adding constraints
# the constraint came from plant 3
model.plant3 = pyo.Constraint(
    expr = 3*model.x1 + 2*model.x2 <= 18
)
model.profit=pyo.Objective(
    expr = 3*model.x1 + 5*model.x2,
    sense = pyo.maximize
)
SOLVER.solve(model)
#model.pprint() 
#model.x1.pprint()
#print(pyo.value(model.x1))# gives it as 1 number.
for o in model.component_objects():# shows it in a better manner, 
    #adding ctype in the bracket above will only give the value of the variables
    print(o, pyo.value(o))
```

```python
# 2. trying out yourself :(
import pyomo.environ as pyo
solver_name= 'appsi_highs'
SOLVER= pyo.SolverFactory(solver_name)
assert SOLVER.available(), f'Solver {solver_name} is not available'

model=pyo.ConcreteModel('Distribution Network')


model.x1 = pyo.Var(bounds=(0,10)) # f1-f2
model.x2 = pyo.Var(bounds=(0,None))
model.x3 = pyo.Var(bounds=(0,None))
model.x4 = pyo.Var(bounds=(0,None))
model.x5 = pyo.Var(bounds=(0,80)) # dc w2
model.x6 = pyo.Var(bounds=(0,None))
model.x7 = pyo.Var(bounds=(0,None))

model.factory1= pyo.Constraint(
    expr = model.x1+model.x2+ model.x3 == 50
)
model.factory2= pyo.Constraint(
    expr = -model.x1+model.x4 == 40
)
model.distribution= pyo.Constraint(
    expr = -model.x2 - model.x4 + model.x5 == 0
)
model.ware1= pyo.Constraint(
    expr = model.x1 + model.x6 - model.x7 == -30
)
model.ware2= pyo.Constraint(
    expr = -model.x5 - model.x6 + model.x7 == -60
)
model.profit=pyo.Objective(
    expr = 2*model.x1 + 4*model.x2 + 9*model.x3+ 3*model.x4 + model.x5 + 3*model.x6 + 2*model.x7,
    sense = pyo.minimize
)
SOLVER.solve(model)
model.pprint() 
```
